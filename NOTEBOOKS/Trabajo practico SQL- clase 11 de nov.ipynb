1. Funciones Matemáticas en Databricks SQL
1. Redondeo y Ajuste Numérico

ROUND(col, n): Redondea al número de decimales especificado n (ideal para publicar índices económicos).

CEIL(col) / CEILING(col): Redondea hacia arriba al entero más cercano.

FLOOR(col): Redondea hacia abajo al entero más cercano.

TRUNC(col, n): Trunca el número dejando n decimales sin aplicar redondeo.

```python
%sql

SELECT 
    pais,
    pib_billones_usd,
    ROUND(pib_billones_usd, 1) AS pib_redondeado,
    CEIL(pib_billones_usd) AS pib_techo,
    FLOOR(pib_billones_usd) AS pib_piso,
    TRUNCATE(pib_billones_usd, 0) AS pib_truncado
FROM workspace.electiva.indicadores_paises;
```

2. Signo y Valor Absoluto

ABS(col): Devuelve el valor absoluto (útil para calcular la brecha o desviación absoluta respecto a una meta de inflación o equilibrio).

SIGN(col): Devuelve 1 si el número es positivo, -1 si es negativo y 0 si es cero (útil para clasificar saldos de balanza comercial o crecimiento/decrecimiento).

```python
%sql
-- Ejemplo: Calcular la desviación absoluta de la inflación respecto a la meta del banco central (3.0%)
SELECT 
    pais,
    tasa_inflacion_pct,
    ABS(tasa_inflacion_pct - 3.00) AS desviacion_meta_inflacion
FROM workspace.electiva.indicadores_paises;
```

3.Aritmética Avanzada, Potencias y Raíces

POWER(base, exp) / POW(base, exp): Eleva la base al exponente especificado (utilizado en modelos de interés compuesto o funciones de producción Cobb-Douglas).

SQRT(col): Calcula la raíz cuadrada (usado en cálculos de volatilidad y modelos de varianza).

MOD(dividendo, divisor) / %: Devuelve el residuo de una división entera.

LN(col) / LOG(col): Logaritmo natural (fundamental en econometría para calcular elasticidades y linealizar tasas de crecimiento).

LOG10(col): Logaritmo en base 10.

```python
%sql
SELECT 
    pais,
    pib_billones_usd,
    LN(pib_billones_usd) AS log_pib, -- Transformación logarítmica econométrica
    POWER(1 + (tasa_inflacion_pct / 100), 2) - 1 AS inflacion_acumulada_2anios,
    LOG10(pib_billones_usd) AS log_pib_base10, --Logaritmo en base 10 del pib_billones_usd,
    EXP(pib_billones_usd) AS exp_pib, --exponencial del pib_billones_usd
    sqrt(pib_billones_usd) AS sqrt_pib --raiz cuadrada del pib_billones_usd
FROM  workspace.electiva.indicadores_paises;
```

3. Funciones de Agregación

Las funciones de agregación toman un conjunto de valores económicos y devuelven un único valor estadístico resumido.

Manejo de Nulos: Las funciones de agregación ignoran los valores NULL, salvo COUNT(*).

COUNT(*): Cuenta el total de países o registros analizados.

COUNT(columna): Cuenta únicamente los países que tienen registros de subsidios (excluye NULLs).

COUNT(DISTINCT columna): Cuenta cuántas regiones económicas distintas existen.

SUM(columna): Suma el producto interno bruto o la población total.

AVG(columna): Calcula la tasa o promedio macroeconómico.

MIN(columna) / MAX(columna): Identifica el valor mínimo o máximo dentro de la muestra.

```python
%sql
SELECT 
    COUNT(*) AS total_paises_analizados,
    COUNT(subsidios_millones) AS paises_con_subsidios_registrados,
    COUNT(DISTINCT region_economica) AS bloques_economicos,
    SUM(pib_billones_usd) AS pib_mundial_acumulado,
    ROUND(AVG(tasa_inflacion_pct), 2) AS inflacion_promedio_global,
    MIN(tasa_inflacion_pct) AS menor_inflacion,
    MAX(tasa_inflacion_pct) AS mayor_inflacion,
    ROUND(STDDEV(tasa_inflacion_pct), 2) AS volatilidad_inflacion
FROM workspace.electiva.indicadores_paises;
```

4. Agrupamiento con GROUP BY y Filtros con HAVING

La cláusula GROUP BY agrupa las economías por región o bloque comercial para comparar métricas macroeconómicas.

WHERE: Filtra los países individuales antes de agrupar.

HAVING: Filtra los bloques económicos después de haber calculado las métricas agregadas.

```python
%sql
SELECT 
    region_economica,
    COUNT(*) AS cantidad_paises,
    ROUND(SUM(pib_billones_usd), 2) AS pib_total_region,
    ROUND(AVG(CAST(tasa_desempleo_pct AS DECIMAL(5,2))), 2) AS desempleo_promedio
FROM workspace.electiva.indicadores_paises
WHERE poblacion_millones >= 50 -- 1. Filtra economías con al menos 50 millones de habitantes
GROUP BY region_economica       -- 2. Agrupa los datos por Región Económica
HAVING AVG(tasa_inflacion_pct) > 3.00 -- 3. Filtra regiones cuya inflación promedio supere el 3%
ORDER BY pib_total_region DESC;
```

```python
%sql
SELECT
    pais,
    ROUND((pib_billones_usd * 1000000000) / (poblacion_millones * 1000000), 2) AS pib_per_capita_usd,
    COALESCE(CAST(subsidios_millones AS DECIMAL(10,2)), 0) AS subsidios_reportados
FROM workspace.electiva.indicadores_paises;
```

```python
%sql
-- 2. Comparativo Macroeconómico por Bloque (GROUP BY + HAVING)
SELECT
    region_economica,
    SUM(poblacion_millones) AS poblacion_total_region,
    MAX(CAST(tasa_desempleo_pct AS DECIMAL(5,2))) AS desempleo_maximo,
    MIN(CAST(tasa_desempleo_pct AS DECIMAL(5,2))) AS desempleo_minimo,
    ROUND(AVG(tasa_inflacion_pct), 2) AS inflacion_promedio
FROM workspace.electiva.indicadores_paises
GROUP BY region_economica
HAVING SUM(pib_billones_usd) >= 3.0 -- Solo regiones con PIB acumulado >= 3.0 billones USD
```
